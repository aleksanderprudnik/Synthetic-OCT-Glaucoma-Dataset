```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# 1. Generate non-linear data (sine wave + noise)
np.random.seed(42)
X = np.sort(np.random.uniform(0, 6, 40)).reshape(-1, 1)
y = np.sin(X).ravel() + np.random.normal(0, 0.25, size=40)
X_test = np.linspace(0, 6, 300).reshape(-1, 1)

# 2. Fit unconstrained, overfitted Decision Tree
tree = DecisionTreeRegressor(max_depth=None, random_state=42)
tree.fit(X, y)
y_tree = tree.predict(X_test)

# 3. Plot noisy points vs overfitted step function
fig, ax = plt.subplots(figsize=(6, 5), dpi=300)
ax.scatter(X, y, color='#2563eb', edgecolors='#1e3a8a', s=45, alpha=0.8, label='Noisy Samples')
ax.plot(X_test, np.sin(X_test), color='#64748b', linestyle=':', linewidth=1.5, label='True Signal: sin(x)')
ax.step(X_test, y_tree, color='#dc2626', linewidth=2, label='Overfitted Tree (High Variance)')

ax.set_title('Decision Tree Overfitting: Sawtooth Effect', fontsize=12, fontweight='bold', color='#0f172a', pad=10)
ax.set_xlabel('Input Feature (X)', fontsize=10, fontweight='bold', color='#475569')
ax.set_ylabel('Target Value (y)', fontsize=10, fontweight='bold', color='#475569')
ax.grid(True, linestyle='--', alpha=0.4)
ax.legend(loc='lower left', fontsize=9)

# Note on variance
ax.text(0.95, 0.95, 'max_depth = None\nMemorizes noise\nZero extrapolation',
        transform=ax.transAxes, fontsize=9, fontweight='bold', color='#b91c1c',
        verticalalignment='top', horizontalalignment='right',
        bbox=dict(boxstyle='round,pad=0.4', facecolor='#fef2f2', edgecolor='#fca5a5'))

plt.tight_layout()
plt.savefig('tree_overfitting_sawtooth.png', dpi=300)
plt.show()
```
